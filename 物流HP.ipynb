```python
import numpy as np
import matplotlib.pyplot as plt
import japanize_matplotlib  # 日本語表示用（必要に応じてインストールしてください）

# xの範囲を設定
x = np.linspace(0, 10, 500)

# 各ルートの一次関数（直線）を定義
# ルートA: 右肩下がり
y_A = -1.2 * x + 9
# ルートB: 緩やかな右肩上がり
y_B = 0.4 * x + 2.6
# ルートC: 急な右肩上がり
y_C = 1.3 * x - 3.7

# 上側包絡線（最大値）を計算
y_max = np.maximum(np.maximum(y_A, y_B), y_C)

# グラフの描画設定
plt.figure(figsize=(10, 6))

# 各ルートの直線を描画（元の図に近い色合い）
plt.plot(x, y_A, label=r'ルートA: $x^T w_A$', color='dodgerblue', linestyle='--', linewidth=2)
plt.plot(x, y_B, label=r'ルートB: $x^T w_B$', color='darkorange', linestyle='--', linewidth=2)
plt.plot(x, y_C, label=r'ルートC: $x^T w_C$', color='forestgreen', linestyle='--', linewidth=2)

# 上側包絡線を太い赤線で描画（ここが修正ポイント）
plt.plot(x, y_max, label=r'$\xi_S(x) = \max_{w \in S} x^T w$', color='red', linewidth=4)

# 最悪ルートが切り替わる点（交点）の計算
x_intersect1 = (9 - 2.6) / (1.2 + 0.4)  # AとBの交点
x_intersect2 = (2.6 - (-3.7)) / (1.3 - 0.4)  # BとCの交点

# 切り替わりポイントの縦線を引く
plt.axvline(x=x_intersect1, color='gray', linestyle=':', linewidth=1.5)
plt.axvline(x=x_intersect2, color='gray', linestyle=':', linewidth=1.5)

# テキストの追加
plt.text(x_intersect1 / 2, 9.5, 'この区間では\nルートAが最大', ha='center', va='center')
plt.text((x_intersect1 + x_intersect2) / 2, 9.5, 'この区間では\nルートBが最大', ha='center', va='center')
plt.text((x_intersect2 + 10) / 2, 9.5, 'この区間では\nルートCが最大', ha='center', va='center')

# 軸の装飾とラベル
plt.ylim(0, 11)
plt.xlim(0, 10)
plt.xlabel(r'$x = c - 2\hat{c}$', fontsize=12)
plt.ylabel('値', fontsize=12)
plt.title('図：各ルートごとの値は直線になる（正確な上側包絡線）', fontsize=14, fontweight='bold')
plt.legend(loc='lower right', fontsize=12)
plt.grid(True, alpha=0.3)

plt.show()
```
